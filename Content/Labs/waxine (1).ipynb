# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

**Answer:** Statement 2

Statement 2 is in our eyes the most probable as when you place the glass over the candle the fire heats the trapped air inside. This leads to thermal expansion of air. As the flame is extinguished, the heated air cools rapidly causing a drop in pressure. The pressure difference between the inside and outside of the glass, draws the water from the outside upwards to restore the balance. The higher atmospheric pressure pushes the water inside. 

Statement 1 is less correct as despite the oxygen being consumed in the process of combustion by definition carbon dioxide and water (H2O vapor) is produced, so this statement does not validly explain how the vacuum is being created as the oxygen depletion is replaced by CO2 and H2O. 

Statement 3 is less correct as visbly by our observations the H2O production was very low, and mostly was CO2 that was being produced, hence the sigificance of the H2O vapor taking space would not be enough to create a large pressure difference to cause the water to rise.
