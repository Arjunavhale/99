# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

De soortelijke warmte $c$ van een stof is de hoeveelheid energie die nodig is om de temperatuur van één kilogram van die stof met één graad Celsius (of één kelvin) te verhogen. De fundamentele relatie wordt gegeven door:

$$
Q= {mc}{ \Delta T}
$$

Daardoor wordt c gegeven door: 

$$
c=\frac{Q}{m \Delta T}
$$

Waarbij: 


Q de totale toegevoerde warmte-energie is in Joules (J), m de massa van de stof is in kilogram (kg), c is de specifieke warmte is in Joules per kilogram per kelvin, (delta)T = T_f - T_i de temperatuurverandering is in graden Celsius of kelvin.

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt

mass_water = 0.250 #kg
expected_c_water = 4186 #theoretical

Voltage = np.array([12.6, 15.0, 16.7, 19.3]) #V
Current = np.array([1.30, 1.55, 1.72, 1.80]) #A
Time = np.array([100, 100, 100, 100,]) #s
Temp_i = 19.4 #deg C - kept temp in degree C because delta T is unitless
Temp_f = np.array([21.0, 21.3, 21.9, 22.8]) #deg C


Power= Voltage * Current #W
Energy = Power * Time #J
Delta_T = Temp_f - Temp_i #deg C
c_water_calculated = Energy / (mass_water * Delta_T) #J/kgK

c_mean=np.mean(c_water_calculated)
print("De gemiddelde soortelijke warmte is:", c_mean)
plt.savefig("soortelijke_warmte_plot.png")
````
